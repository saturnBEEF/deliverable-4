# Task 1: Testing the Temperature class

The tests are written using pytest and cover:
- Default unit, case-insensitive units and read-only attributes.
- Invalid units and temperatures below absolute zero.
- Conversions between Celsius, Fahrenheit and Kelvin at five reference points.
- Round-trip conversion from Celsius to Fahrenheit to Kelvin and back.
- Equality and ordering across different units.
- String representations using str() and repr().

Floating-point results are checked using pytest.approx with an absolute
tolerance of 1e-9 and no relative tolerance.
